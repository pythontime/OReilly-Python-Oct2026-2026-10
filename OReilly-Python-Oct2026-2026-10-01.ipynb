# How and where can you write Python code?

If you're going to write Python, you're going to need two things: (a) a copy of the Python language and (b) an editor that's familiar with Python, and lets you create files -- and probably run them.

Traditionally, you would do this by downloading Python from python.org, and then installing it, and then installing an editor -- such as PyCharm or Visual Studio Code. 

There is, however, an alternative called Jupyter that has become massively popular. I'm typing into Jupyter now! It's known as a "notebook," and it's particularly popular in the data community. You can install Jupyter on your own computer, once you've installed Python.

Jupyter gives us a highly interactive enviroment in which we can write code, see the reuslts of our code right away, and iterate quickly.

But... installing it is a bear if you're new. Installing Python is easier than used to be the case, but it's also annoying.

# Intro to using Jupyter

Jupyter works on a system of "cells." Right now, I'm typing into a cell. Each cell at a given moment, can be in one of two modes (when I'm typing into it):

- Edit mode. When I'm in edit mode, typing enters characters, like this. I can get into edit mode by clicking in the cell or by pressing Enter.
- Command mode. When I'm in command mode, typing a character doesn't make it visible. Rather, it tells Jupyter to do something. I can get into command mode by clicking to the left of the cell or by pressing ESC.
