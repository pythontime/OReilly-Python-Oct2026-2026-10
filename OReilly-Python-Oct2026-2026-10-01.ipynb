# Intro to using Jupyter

Jupyter works on a system of "cells." Right now, I'm typing into a cell. Each cell at a given moment, can be in one of two modes (when I'm typing into it):

- Edit mode. When I'm in edit mode, typing enters characters, like this. I can get into edit mode by clicking in the cell or by pressing Enter.
- Command mode. When I'm in command mode, typing a character doesn't make it visible. Rather, it tells Jupyter to do something. I can get into command mode by clicking to the left of the cell or by pressing ESC.

A few commands:

- `c` -- copies the cell
- `x` -- cuts the cell
- `v` -- pastes the most recently copied/cut cell in the notebook
- `a` -- add a new cell *above*
- `b` -- add a new cell *below*
- `m` -- put the cell into Markdown (documentation) mode
- `y` -- put the cell into coding (Python) mode

When I'm done editing a cell in edit mode, I can press shift+ENTER together, and that "executes" the cell, meaning that the code runs.

# What is a programming language?

Back when computers were invented, they were collections of vacuum tubes, wires, etc. If you wanted to solve a problem, you built a computer to do so. If you wanted to solve a different problem, you built a separate computer to do so. This was clearly very inefficient.

At some point, people realized that it would be better to build a computer a single time, and solve many different problems. This was the birth of the *programmable computer*. You would program it using 1s and 0s. That worked, but was really tedious and annoying.

That led to the creation of high-level programming languages. You would write in something resembling English, and a second program (a language) would translate your code into 1s and 0s.

That is largely how programming languages work today! You write your code, it is translated into the 1s and 0s, and it is executed.

There are many *many* different programming languages out there. Each has its own advantages and disadvantages.

- C -- runs very very fast, because it's so similar to the 1s and 0s. However, it's hard for people to work with, and the coders need to keep track of how much memory has been used.
- Rust -- a very new language, competing with C, trying to make it easier to write code that runs fast and doesn't require lots of worrying about memory
- Java -- an even higher level language, which now runs super fast, but which took some time to take off and become popular. It uses a style of programming known as "objects," which we won't really be talking about in this course.
- Python -- a very high level language, meaning that it looks roughly like English in many places. The ideaw of Python is to make it easy for someone to learn to program and then to actually code.

# `=` is not equality!

This one-character symbol doesn't mean the same thing as you learned in school. Here, it's just a way of assigning the value on the right to the variable on the left.

```python
x = 10
y = 20

print(x + y)
```

# Why no declarations?

In many other langugages, when (or before) I assign to a variable, we must declare the variable to be of a certain type -- number, text, etc.

This doesn't really exist in Python.

The reason you need this in other languages is because they need to know how much memory to allocate. In those languages, a variable is basically an alias for a location in memory. If the location isn't the right size for the data, bad news.

Any variable in Python can refer to any value. So there's no need for that declaration.

# A few things to notice

1. In Python, you *can* use capital letters in variable names, but most people don't. So use `first_name`, not `First_name`. THESE ARE DIFFERENT VARIABLES! If you use a name capitalized (e.g., `X`) and you defined it with lowercase, then there won't be a match.
2. This is true for Python's functions, too! If you call `Print` or `PRINT`, neither will work. It has to be lowercase `print`.
3. Variable names (into which we assign) have a few rules and a few conventions:
    - We normally only use lowercase letters
    - If we need multiple words, we use `_` between the words
    - You can use lowercase letters, uppercase letters (but try to avoid them), `_`, and digits.
    - Don't start or end a variable name with `_`
    - Don't start a variable name with a digit.
4. There is a world of difference between `x` (the variable) and `'x'` (the letter x).
